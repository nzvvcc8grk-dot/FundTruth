## 1. The problem: fund claims are cheap to make and expensive to check
Every fund makes two kinds of promise. **Numbers**: the returns and the risk it reports. **Words**: its style, its
benchmark and, increasingly, its ESG exclusions. Advisers, fund selectors and investors rely on both, but almost nobody
tests them. Checking today means reading factsheets by hand or buying ESG ratings that measure something else.

**Failure 1: returns that are too smooth or too good.** Funds holding illiquid assets (property, private credit, hedge
strategies) are valued by appraisal or by model. Stale or managed valuations make returns look smoother and safer than
the assets really are (Getmansky, Lo & Makarov, 2004). Managers also avoid reporting small losses (Bollen & Pool, 2009).
Elsewhere, "active" funds charge active fees while hugging an index (Cremers & Petajisto, 2009). The extreme case,
Madoff, was famous for an implausibly smooth track record.

**Failure 2: ESG promises the portfolio does not keep.** Greenwashing is now an enforcement issue. The SEC fined DWS
**US&#36;19 million** in September 2023 for misleading ESG statements, its largest greenwashing penalty, after fines on Goldman
Sachs Asset Management (US&#36;4m, 2022) and BNY Mellon (US&#36;1.5m, 2022). ESG ratings do not settle the question: providers
disagree widely (Berg, Kölbel & Rigobon, 2022), and a rating says how "green" a fund looks, not whether it does what its
own documents promise.

### Why now: in the UK and EU, ESG claims have become a matter of proof
| 31 Jul 2023 | 31 May 2024 | 21 Nov 2024 | 2 Dec 2024 | 21 May 2025 |
|---|---|---|---|---|
| FCA Consumer Duty: firms must evidence fair value for the products they recommend | FCA anti-greenwashing rule: sustainability claims must be fair, clear and capable of being substantiated | ESMA guidelines on fund names apply to new EU funds using ESG terms | FCA SDR naming and marketing rules for UK funds | ESMA fund-name guidelines extend to existing EU funds |

### A large market that has to document its checks
- **US&#36;3.51 trillion** in sustainable funds at the end of March 2026, about 85% of it in Europe (Morningstar).
- **£11.1 trillion** managed in the UK in 2025, a record (Investment Association).
- **31,000 UK financial advisers** in about 5,500 firms, advising on £1 trillion of client assets (FCA, 2025 survey).
  Every one of them must show that the funds they recommend are what they claim to be.

**Our answer in one line:** a tool that reads what a fund *says*, measures what it *does*, and turns every gap into a
precise, polite **question for the manager**: never an accusation, always a documented reason to ask.

## 2. The product: a verdict, the evidence and a question for the manager
FundTruth is a web app on a Python engine. A user picks a fund from the watchlist or adds their own, and in seconds
receives a plain-English verdict, the numbers behind each check, and the questions to send to the manager.

| 1. Inputs | 2. Nine checks in two modules | 3. Output |
|---|---|---|
| **Public data**: monthly prices, market factors, issuer holdings files, the fund's own documents (KID, factsheet). **Your data**: paste from Excel or a factsheet, or upload a file; units and dates are recognised automatically. | **Integrity of returns**: Price regularity, Style and benchmark, Return beyond the market, Small gains and losses, Position among peers (machine learning), Shape of risk. **ESG promises**: Promise vs portfolio, Promise vs behaviour, Label vs index. No ESG ratings: the fund's own words are the benchmark. | Status per check (Nothing unusual, Worth a look, Unusual), evidence and peer range, **questions for the manager**, a one-page PDF report, and the option to add the fund to the comparison set. |

![Fund card](assets/fund_card_SPYX.png)
<p class="muted">The fund card (SPYX): the verdict comes first, then six headline numbers (including the yearly cost of
the ESG label), then the questions for the manager. The checks sit in tabs below.</p>

**User journey.** (1) **Screen**: the watchlist ranks 48 funds by the number of open questions. (2) **Inspect**: the fund
card shows the verdict, the evidence behind each check, and what would make the signal innocent. (3) **Check your own
fund**: paste a factsheet's return table or upload a file; three simple questions pick the right factors and peers.
(4) **Document**: export the one-page PDF, send the questions to the manager and file the answer, an audit trail for
Consumer Duty and SDR.

| User | Job to be done |
|---|---|
| Fund selectors, wealth managers | Screen a fund range and document due diligence before a fund goes on the buy list |
| Independent financial advisers | Show the FCA that recommended funds are what they claim to be, at a fraction of the cost of a research platform |
| Compliance teams at asset managers | Test their own ESG marketing against their own portfolios before a regulator or journalist does |
| Fund platforms | Monitor every fund on the shelf through an API and flag the ones that need a question |

**What makes it different.** *Questions, not scores*: each signal becomes a concrete question, for example "which holdings
does the ESG screen actually remove, and how is the extra fee justified?". *Promise-based, not rating-based*: the fund is
tested against its own binding words. *Two lenses in one place*: return integrity and ESG claims are usually sold
separately. *Built for illiquid funds*: the smoothing and hidden-loss tests target property, private-credit and hedge
funds. *Private by design*: return files stay on the user's computer; only anonymous summary statistics can join the
comparison set.

## 3. Method and data: nine checks, each built on a published method
A status needs statistical evidence: **Unusual** if p < 0.05, **Worth a look** if p < 0.10, otherwise **Nothing unusual**.
P-values are empirical, against a **fixed reference set** of 40 real funds that changes only when a user adds a fund.

| Check | Question | Method | Basis |
|---|---|---|---|
| Price regularity | Are returns too smooth for market prices? | Lag 1-3 autocorrelation, Ljung-Box test, rank against peers; AR(1) de-smoothing | Getmansky, Lo & Makarov (2004) |
| Style and benchmark | Is an "active" fund an index tracker in disguise? | Fama-French 3-factor regression; R² ≥ 0.95 with low tracking error and an above-median fee | Cremers & Petajisto (2009) |
| Return beyond the market | Is the alpha too good to be true? | Annualised alpha, t ≥ 2 and upper tail against peers; sub-periods (2020, 2022-23) | Fama & French (1993) |
| Small gains and losses | Are small losses being hidden? | Binomial test: returns just above vs just below zero | Bollen & Pool (2009) |
| Position among peers | Does the risk profile look unlike similar funds? | Isolation Forest on volatility, skewness, autocorrelation, alpha and R², plus k-means distance | Liu, Ting & Zhou (2008) |
| Shape of risk | What does the risk look like? | Drawdown, skewness, kurtosis, up- and down-capture | Descriptive only |
| Promise vs portfolio | Does the fund hold what it says it excludes? | Holdings against the parent index, the energy sector and Norway's exclusion list, matched to the fund's own wording | Holdings files; NBIM list |
| Promise vs behaviour | Does a "fossil-free" fund still move with oil and gas? | Regression of the fund's return over its index fund on energy's return over the same fund, Newey-West errors | Newey & West (1987) |
| Label vs index | Is the ESG version really different, and what does it cost? | R², tracking error, active share; fee gap in £ per £10,000 a year and over 20 years | Cremers & Petajisto (2009) |

**Two key formulas.** Smoothed reporting, with R* the reported return: R*ₜ = θ₀Rₜ + θ₁Rₜ₋₁ + θ₂Rₜ₋₂ and
θ₀ + θ₁ + θ₂ = 1; the de-smoothed return is (R*ₜ − ρR*ₜ₋₁) / (1 − ρ), with ρ the lag-1 autocorrelation. Promise vs
behaviour: r(ESG) − r(parent) = a + b·[r(energy) − r(parent)] + ε, implied energy underweight w = −b / (1 − b) ≈ −b.

**Data, public and free.** Yahoo Finance monthly total returns (Jan 2016 to Sep 2026); Kenneth R. French Data Library
(US factors); issuer holdings files and fund pages; Norges Bank exclusion list (via OpenSanctions). Universe: 40 liquid US
funds and ETFs (16 equity, 14 bond, 10 multi-asset) plus 8 ESG funds compared with their ordinary index fund.

**Engineering.** Frozen statistical engine protected by a hash test; 161 automated tests; key numbers recomputed by
separate code. Python (pandas, statsmodels, scikit-learn), Streamlit, ReportLab. The code below reads the results the app
produces for all 48 funds:

```python
import pandas as pd
R = "../app/results/"                                   # results produced by the FundTruth engine
ov = pd.read_csv(R + "overview.csv")
print(f"Funds screened: {len(ov)}, of which ESG: {ov['ESG fund'].sum()}")
print(f"Funds with at least one question for the manager: {(ov['Questions for the manager'] > 0).sum()}")
q = ov[ov["Questions for the manager"] > 0]
print(); print(q[["Fund", "Flagged checks"]].to_string(index=False))
```

## 4. Proof that it works
No public dataset labels which funds misreport, so we test FundTruth the way a laboratory tests a diagnostic: we
**plant known anomalies** in real funds (positive control), count **false alarms** on simulated honest funds, run it on
**liquid funds** where little should be found (negative control), and examine a **real ESG case** with two methods.

- **Positive control on real funds.** We distorted the returns of all 40 liquid funds: strong smoothing was found in
  **40 of 40**, mild smoothing in 13 of 40 and hidden small losses in **37 of 40**.
- **False alarms.** On simulated honest funds each check flags on average only **2.5-4.2%** by mistake at the 5% level.
- **Negative control.** On 40 liquid, daily-priced US funds only 3 (EMB, FDGRX, TLT) get a "Worth a look" and **none** is
  "Unusual". Honest simulated funds run through the same pipeline reach "Unusual" 2% of the time (300 funds).
- **ESG module, 8 real funds.** Four keep their promises on all the data that apply (ESGV, SUSA, SUSW.L, V3AB.L); four
  raise one question each (SPYX, CRBN, EFIV, ESGU).

```python
import matplotlib.pyplot as plt
nc = pd.read_csv(R + "negative_control.csv").set_index(["category", "pricing"])
esg = pd.read_csv(R + "esg/esg_summary.csv").set_index("fund").loc["SPYX"]
lo, mid, hi = (esg[c] * 100 for c in ["fossil_ci_low", "fossil_implied_underweight",
                                       "fossil_ci_high"])
held = (esg.parent_energy_weight - esg.energy_weight) * 100      # from holdings files
fig, ax = plt.subplots(1, 2, figsize=(9.2, 1.4), gridspec_kw={"width_ratios": [1.15, 1]})
ax[0].barh(["Strong smoothing", "Hidden small losses", "Mild smoothing"],
           [100, 92.5, 32.5], color="#0B2A4A")
ax[0].invert_yaxis(); ax[0].set_title("Found when planted in 40 real funds (%)", fontsize=9)
ax[1].errorbar([mid], [1], xerr=[[mid - lo], [hi - mid]], fmt="o", color="#0B2A4A", capsize=4)
ax[1].plot([held], [0], "D", color="#AD6800")
ax[1].set_yticks([0, 1], ["From holdings", "From returns"]); ax[1].set_xlim(0, 4)
ax[1].set_ylim(-0.6, 1.6); ax[1].set_title("SPYX energy underweight vs SPY (pp)", fontsize=9)
for a in ax:
    a.spines[["top", "right"]].set_visible(False); a.tick_params(labelsize=8)
plt.tight_layout(); plt.show()
u = nc.loc[("all", "all")]
print(f"Honest simulated funds reaching 'Unusual': {u.unusual_pct:.0f}% of {u.funds:.0f}")
print(f"SPYX: returns imply {mid:.1f}% less energy ({lo:.1f} to {hi:.1f}, "
      f"t = {esg.fossil_t:.1f}); holdings give {held:.1f}%")
```

**Case study: SPDR S&P 500 Fossil Fuel Reserves Free ETF (SPYX).** Promise: *"eliminating companies that own fossil
fuel reserves from the S&P 500"*. **Promise vs behaviour: Nothing unusual.** Over 129 months SPYX's return over SPY falls
when energy rises (slope -0.027, t = -7.5): a portfolio holding 2.6% less energy; the holdings give 2.4%, so two
independent methods are consistent. **Promise vs portfolio: Nothing unusual.** The energy companies still held (refiners,
pipelines) and 12 coal-burning utilities (0.60%) do not own reserves. **Label vs index: Worth a look.** R² 99.3% with SPY,
tracking error 1.25%, active share 3%; the fee is 0.20% against 0.0945%, **£11 a year per £10,000**, about £632 over 20
years. **The question FundTruth sends:** *which holdings does the ESG screen actually remove, what share of SPY is that,
and how is the extra fee justified?* The promise is kept; what needs explaining is the price of the label.

**Where module 1 earns its keep.** A property fund valued by appraisal (simulated example uploaded through "Check your
own fund") reads **Unusual** on Price regularity, and the question is about the valuation policy.

**What the evidence says, honestly.** On liquid funds module 1 finds little, as it should; its value is in illiquid funds.
Same-type groups are still small (10-16 funds), so each fund is compared with all 40 and a few signals will be chance:
hence a question, not a verdict. CRBN's question comes from a weak energy underweight (t = -1.0). Holdings exist for 2 of
8 ESG funds; US factors only (no momentum), so SUSW.L and V3AB.L skip the factor checks; thresholds not tuned on test data.

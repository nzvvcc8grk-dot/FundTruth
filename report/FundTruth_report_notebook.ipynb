<p class="lab"><b>1</b>The problem</p><h2 class="t">Fund claims are cheap to make and expensive to check</h2><p class="lead">Every fund makes two kinds of promise. <b>Numbers</b>: the returns and the risk it reports. <b>Words</b>: its style, its benchmark and, increasingly, its ESG exclusions. Advisers, fund selectors and investors rely on both, but almost nobody tests them. Checking today means reading factsheets by hand or buying ESG ratings that measure something else.</p>
<div class="g2">
<div class="mod"><h3>Failure 1: returns that are too smooth or too good</h3><p>Funds holding illiquid assets (property, private
credit, hedge strategies) are valued by appraisal or by model. Stale or managed valuations make returns look smoother and
safer than the assets really are (Getmansky, Lo &amp; Makarov, 2004). Managers also avoid reporting small losses (Bollen &amp;
Pool, 2009). Elsewhere, "active" funds charge active fees while hugging an index (Cremers &amp; Petajisto, 2009). The extreme
case, Madoff, was famous for an implausibly smooth track record.</p></div>
<div class="mod"><h3>Failure 2: ESG promises the portfolio does not keep</h3><p>Greenwashing is now an enforcement issue. The
SEC fined DWS <b>US&#36;19 million</b> in September 2023 for misleading ESG statements, its largest greenwashing penalty, after
fines on Goldman Sachs Asset Management (US&#36;4m, 2022) and BNY Mellon (US&#36;1.5m, 2022). ESG ratings do not settle the
question: providers disagree widely (Berg, Kölbel &amp; Rigobon, 2022), and a rating says how "green" a fund looks, not
whether it does what its own documents promise.</p></div></div>
<h3 class="mt">Why now: in the UK and EU, ESG claims have become a matter of proof</h3>
<div class="timeline">
<div><b>31 Jul 2023</b>FCA Consumer Duty: firms must evidence fair value for the products they recommend</div>
<div><b>31 May 2024</b>FCA anti-greenwashing rule: sustainability claims must be fair, clear and capable of being substantiated</div>
<div><b>21 Nov 2024</b>ESMA guidelines on fund names apply to new EU funds using ESG terms</div>
<div><b>2 Dec 2024</b>FCA SDR naming and marketing rules for UK funds</div>
<div><b>21 May 2025</b>ESMA fund-name guidelines extend to existing EU funds</div></div>

<p class="lab"><b>3</b>Method and data</p><h2 class="t">Nine checks, each built on a published method and free public data</h2><p class="lead">A status needs statistical evidence: <span class="st bad">Unusual</span> if p &lt; 0.05, <span class="st look">Worth a look</span> if p &lt; 0.10, otherwise <span class="st ok">Nothing unusual</span>. P-values are empirical, against a <b>fixed reference set</b> of 40 real funds that changes only when a user adds a fund.</p>
<table>
<tr><th style="width:15%">Check</th><th style="width:24%">Question</th><th>Method</th><th style="width:17%">Basis</th></tr>
<tr class="grp"><td colspan="4">Module 1: Integrity of returns</td></tr>
<tr><td>Price regularity</td><td>Are returns too smooth for market prices?</td><td>Lag 1-3 autocorrelation, Ljung-Box test, rank against peers; AR(1) de-smoothing</td><td>Getmansky, Lo &amp; Makarov (2004)</td></tr>
<tr><td>Style and benchmark</td><td>Is an "active" fund an index tracker in disguise?</td><td>Fama-French 3-factor regression; R² ≥ 0.95 with low tracking error and an above-median fee</td><td>Cremers &amp; Petajisto (2009)</td></tr>
<tr><td>Return beyond the market</td><td>Is the alpha too good to be true?</td><td>Annualised alpha, t ≥ 2 and upper tail against peers; sub-periods (2020, 2022-23)</td><td>Fama &amp; French (1993)</td></tr>
<tr><td>Small gains and losses</td><td>Are small losses being hidden?</td><td>Binomial test: returns just above vs just below zero</td><td>Bollen &amp; Pool (2009)</td></tr>
<tr><td>Position among peers</td><td>Does the risk profile look unlike similar funds?</td><td>Isolation Forest on volatility, skewness, autocorrelation, alpha and R², plus k-means distance</td><td>Liu, Ting &amp; Zhou (2008)</td></tr>
<tr><td>Shape of risk</td><td>What does the risk look like?</td><td>Drawdown, skewness, kurtosis, up- and down-capture</td><td>Descriptive only</td></tr>
<tr class="grp"><td colspan="4">Module 2: ESG promises</td></tr>
<tr><td>Promise vs portfolio</td><td>Does the fund hold what it says it excludes?</td><td>Holdings against the parent index, the energy sector and Norway's exclusion list, matched to the fund's own wording</td><td>Holdings files; NBIM list</td></tr>
<tr><td>Promise vs behaviour</td><td>Does a "fossil-free" fund still move with oil and gas?</td><td>Regression of the fund's return over its index fund on energy's return over the same fund, Newey-West errors</td><td>Newey &amp; West (1987)</td></tr>
<tr><td>Label vs index</td><td>Is the ESG version really different, and what does it cost?</td><td>R², tracking error, active share; fee gap in £ per £10,000 a year and over 20 years</td><td>Cremers &amp; Petajisto (2009)</td></tr>
</table>
<div class="g2 mt">
<div class="mod"><h3>Two key formulas</h3><p>Smoothed reporting, with R* the reported return:</p>
<div class="eq">R*ₜ = θ₀Rₜ + θ₁Rₜ₋₁ + θ₂Rₜ₋₂, &nbsp;θ₀ + θ₁ + θ₂ = 1<br>de-smoothed: (R*ₜ − ρR*ₜ₋₁) / (1 − ρ), ρ = lag-1 autocorrelation</div>
<p>Promise vs behaviour: the slope estimates the missing energy weight.</p>
<div class="eq">r(ESG) − r(parent) = a + b·[r(energy) − r(parent)] + ε<br>implied energy underweight: w = −b / (1 − b) ≈ −b</div></div>

separate code. Python (pandas, statsmodels, scikit-learn), Streamlit, ReportLab.</li></ul>

<p class="lab"><b>4</b>Evidence</p><h2 class="t">FundTruth finds planted problems and rarely flags honest funds</h2><p class="lead">No public dataset labels which funds misreport, so we test FundTruth the way a laboratory tests a diagnostic: we <b>plant known anomalies</b> in real funds (positive control), count <b>false alarms</b> on simulated honest funds, run it on <b>liquid funds</b> where little should be found (negative control), and examine a <b>real ESG case</b>.</p>
<div class="g2"><ul>
<li><b>Positive control.</b> We distorted the returns of all 40 liquid funds: strong smoothing was found in <b>40 of 40</b>,
mild smoothing in 13 of 40, hidden small losses in <b>37 of 40</b>.</li>
<li><b>False alarms.</b> On simulated honest funds each check flags on average only <b>2.5-4.2%</b> by mistake at the 5% level.</li></ul>
<ul><li><b>Negative control.</b> Of 40 liquid US funds, only 3 (EMB, FDGRX, TLT) get "Worth a look" and <b>none</b> is
"Unusual"; honest simulated funds reach "Unusual" 2% of the time (300 funds).</li>
<li><b>ESG module, 8 real funds.</b> Four keep their promises on all the data that apply; four raise one question each
(SPYX, CRBN, EFIV, ESGU).</li></ul></div></div>

```python
nc = pd.read_csv(R + "negative_control.csv").set_index(["category", "pricing"])
esg = pd.read_csv(R + "esg/esg_summary.csv").set_index("fund").loc["SPYX"]
lo, mid, hi = (esg[c] * 100 for c in ["fossil_ci_low", "fossil_implied_underweight", "fossil_ci_high"])
held = (esg.parent_energy_weight - esg.energy_weight) * 100          # from the holdings files
fig, ax = plt.subplots(1, 2, figsize=(9.4, 1.45), gridspec_kw={"width_ratios": [1.2, 1]})
labels, vals = ["Strong smoothing", "Hidden small losses", "Mild smoothing"], [100, 92.5, 32.5]
ax[0].barh(labels, vals, color=NAVY, height=0.55); ax[0].invert_yaxis(); ax[0].set_xlim(0, 118)
for i, v in enumerate(vals): ax[0].text(v + 2, i, f"{v:g}%", va="center", color=NAVY, fontweight="bold")
ax[0].set_title("Found when planted in 40 real funds", loc="left", fontweight="bold")
ax[1].errorbar([mid], [1], xerr=[[mid - lo], [hi - mid]], fmt="o", color=NAVY, capsize=4)
ax[1].plot([held], [0], "D", color=AMBER); ax[1].set_xlim(0, 4); ax[1].set_ylim(-0.6, 1.6)
ax[1].set_yticks([0, 1], ["From holdings", "From returns"])
ax[1].set_title("SPYX: energy underweight vs SPY (pp)", loc="left", fontweight="bold")
plt.tight_layout(); plt.show()
u = nc.loc[("all", "all")]
print(f"Honest simulated funds flagged 'Unusual': {u.unusual_pct:.0f}% of {u.funds:.0f}")
print(f"SPYX underweight: {mid:.1f}% from returns ({lo:.1f} to {hi:.1f}, t = {esg.fossil_t:.1f}), "
      f"{held:.1f}% from holdings")
```
